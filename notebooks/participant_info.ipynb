```python
import pandas as pd
from pathlib import Path

bids_folder = Path('/data/ds-neuralpriors')

df = pd.read_csv(bids_folder / 'participants.tsv', sep='\t')

# Number of females, mean age, age range
num_females = df[df['Sex'] == 'F'].shape[0]
mean_age = df['age'].mean()
age_range = df['age'].min(), df['age'].max()
# Print the results
print('Total number of participants:', df.shape[0])
print(f"Number of females: {num_females}")
print(f"Mean age: {mean_age}")
print(f"Age range: {age_range}")
```

One-shot import from the recruitment export (`~/Downloads/participants.csv`) to `participants.tsv` — do not re-run.

```python
import pandas as pd


from neural_priors.utils.data import get_all_subject_ids

subject_ids = [int(s) for s in get_all_subject_ids()]

fn = '/Users/gdehol/Downloads/participants.csv'
df = pd.read_csv(fn, header=1, usecols=[0, 1,2,3], index_col=0)

# Drop the first row
df = df.iloc[1:]

df.index.name = 'participant_id'

# Drop all row where index is nan
df = df.loc[df.index.notna()]

df.index = df.index.astype(int)
df = df.loc[subject_ids]
df = df.rename(columns={
    'Age':'age', 'Gender':'Sex'}).drop('Name', axis=1)


df.to_csv('/data/ds-neuralpriors/participants.tsv', sep='\t')
```
