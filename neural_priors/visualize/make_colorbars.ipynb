```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import Normalize, ListedColormap
from matplotlib.colorbar import ColorbarBase
from pathlib import Path

plt.rcParams['font.family'] = 'Helvetica'

# Set up the figure and axis (slightly higher)
fig, ax = plt.subplots(figsize=(6, 0.6))
fig.subplots_adjust(bottom=0.5)

# Create a custom colormap: black for [0, 0.01), inferno for [0.01, 0.05]
base_cmap = plt.get_cmap('inferno')
colors = base_cmap(np.linspace(0, 1, 256))
threshold_idx = int(0.01 / 0.05 * 255)
colors[:threshold_idx] = [0, 0, 0, 1]  # set to black
custom_cmap = ListedColormap(colors)

norm = Normalize(vmin=0.0, vmax=0.05)

# Create colorbar with larger font size, Helvetica font, no label
cb = ColorbarBase(ax, cmap=custom_cmap, norm=norm, orientation='horizontal')
cb.ax.tick_params(labelsize=22, labelcolor='black')
for label in cb.ax.get_xticklabels():
    label.set_fontname('Helvetica')

# plt.show()

bids_folder = Path('/data/ds-neuralpriors')
plt.savefig(bids_folder / 'derivatives' / 'figures' / 'colorbar_r2.pdf', bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import Normalize, ListedColormap
from matplotlib.colorbar import ColorbarBase

plt.rcParams['font.family'] = 'Helvetica'

# Set up the figure and axis (same height as previous)
fig, ax = plt.subplots(figsize=(6, 0.6))
fig.subplots_adjust(bottom=0.5)

# Create a custom colormap: black for [5, 7), nipy_spectral for [7, 25]
base_cmap = plt.get_cmap('nipy_spectral')
colors = base_cmap(np.linspace(0, 1, 256))
threshold_idx = int((7-5) / (25-5) * 255)
colors[:threshold_idx] = [0, 0, 0, 1]  # set to black
custom_cmap = ListedColormap(colors)

norm = Normalize(vmin=5, vmax=25)

# Create colorbar with larger font size, Helvetica font, no label
cb = ColorbarBase(ax, cmap=custom_cmap, norm=norm, orientation='horizontal')
cb.ax.tick_params(labelsize=22, labelcolor='black')
for label in cb.ax.get_xticklabels():
    label.set_fontname('Helvetica')

plt.savefig(bids_folder / 'derivatives' / 'figures' / 'colorbar_mu.pdf', bbox_inches='tight')
```
